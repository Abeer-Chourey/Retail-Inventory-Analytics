```python
# Create useful date features

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()
df["Quarter"] = df["Date"].dt.quarter
df["Day_of_Week"] = df["Date"].dt.dayofweek

df[
    [
        "Date",
        "Year",
        "Month",
        "Month_Name",
        "Week",
        "Day_Name",
        "Quarter"
    ]
].head()
```

## Initial Business Analysis

```python
# Question 1: Total units sold

total_units_sold = df["Units Sold"].sum()

print("Total Units Sold:", total_units_sold)
```

```python
# Question 2: Total demand

total_demand = df["Demand"].sum()

print("Total Demand:", total_demand)
```

```python
# Question 3: Average inventory level

average_inventory = df["Inventory Level"].mean()

print("Average Inventory Level:", round(average_inventory, 2))
```

```python
# Question 4: Which categories sell the most?

category_sales = (
    df.groupby("Category")["Units Sold"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
# Visualize category sales

plt.figure(figsize=(9, 5))

category_sales.plot(kind="bar")

plt.title("Units Sold by Category")
plt.xlabel("Category")
plt.ylabel("Units Sold")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Question 5: Which stores sell the most?

store_sales = (
    df.groupby("Store ID")["Units Sold"]
    .sum()
    .sort_values(ascending=False)
)

store_sales
```

```python
# Visualize store sales

plt.figure(figsize=(8, 5))

store_sales.plot(kind="bar")

plt.title("Units Sold by Store")
plt.xlabel("Store ID")
plt.ylabel("Units Sold")

plt.tight_layout()
plt.show()
```

```python
# Question 6: Which products have the highest demand?

product_demand = (
    df.groupby("Product ID")["Demand"]
    .sum()
    .sort_values(ascending=False)
)

product_demand.head(10)
```

```python
# Question 7: Inventory vs Sales

inventory_sales = (
    df.groupby("Product ID")
    .agg(
        Average_Inventory=("Inventory Level", "mean"),
        Total_Units_Sold=("Units Sold", "sum"),
        Total_Demand=("Demand", "sum")
    )
    .sort_values("Total_Demand", ascending=False)
)

inventory_sales.head(10)
```

```python
# Exploratory inventory-to-demand ratio

inventory_sales["Inventory_to_Demand_Ratio"] = (
    inventory_sales["Average_Inventory"] /
    inventory_sales["Total_Demand"]
)

inventory_sales.head()
```
